## Stemming

### Porter

# First dividing by word then rejoining into a sentences

```python
# words
for index, row in df.iterrows():
  stemmed_words = []

  for word in row['tokenized_text_words']:
     if word not in stop_words:
      try:
        stemmed_word = stemmer_ps.stem(word)
        stemmed_words.append(stemmed_word)
      except RecursionError:
        print(f'RecursionError in row {index} on word: {word}')
        continue
  df.at[index, 'porter_stem_words'] = stemmed_words
```

```python
# stemming

for index, row in df.iterrows():
    stemmed_sents = []

    for sent in row['tokenized_text_sents']:
        # Split sentence into words
        words = sent.split()

        # Stem each word individually
        stemmed_words_in_sent = []
        for word in words:
            try:
                stemmed_word = stemmer_ps.stem(word)
                stemmed_words_in_sent.append(stemmed_word)
            except RecursionError:
                print(f'RecursionError in row {index} on word: {word}')
                # Use original word if stemming fails
                stemmed_words_in_sent.append(word)

        # Rejoin the stemmed words into a sentence
        stemmed_sent = ' '.join(stemmed_words_in_sent)
        stemmed_sents.append(stemmed_sent)

    # Assign the stemmed sentences to the current row
    df.at[index, 'porter_stem_sents'] = stemmed_sents
```
