```python
# Drop rows where label is 'idk/skip' or 'relation'
vicom_labels = vicom_labels[~vicom_labels['label'].isin(['idk/skip', 'relation'])]

# Verify the labels are now only 'noHate' and 'hate'
print(vicom_labels['label'].unique())
```

```python
import numpy as np
import os

vicom_labels['num_labels'] = np.nan

vicom_files = "/home/sam/Projects/Senior Capstone/CapstoneProject/Data/vicom_data/all_files"

for file in os.listdir(vicom_files):
    filename = os.fsdecode(file)

    filename, ext = os.path.splitext(filename)

    # find rows where file_id == filename
    mask = (vicom_labels['file_id'] == filename)

    if mask.any():
        label = vicom_labels.loc[mask, 'label'].iloc[0]

        if label == 'noHate':
            vicom_labels.loc[mask, 'num_labels'] = 0
        elif label == 'hate':
            vicom_labels.loc[mask, 'num_labels'] = 1

```

```python
vicom_labels['num_labels'] = vicom_labels['num_labels'].astype('int64')
```
